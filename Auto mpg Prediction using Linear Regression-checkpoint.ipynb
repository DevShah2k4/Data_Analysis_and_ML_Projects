```python
#here the mapping of dictionary is done 
d1 = {1:"USA",2:"EUROPE",3:"ASIA"}
auto["origin"].replace(d1,inplace=True)
```

```python
auto["horsepower"]= pd.to_numeric(auto["horsepower"],errors="coerce")
auto["horsepower"] = auto["horsepower"].fillna(auto["horsepower"].mean())
```

```python
#here for the prediction the mpg is target variable and the cyclinders,displacement,weight,horsepower are independent variable taken into x&y respectively
x = auto[["cylinders","displacement","weight","horsepower","acceleration"]]
y = auto["mpg"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
#here we have done the training and testing and spliting the data into 70:30 
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
#here regression is the object
regression = LinearRegression()
regression.fit(x_train,y_train)
```

```python
#here prediction is done on the basis of the four fetaures cyclinder,displacement,weight,horsepower,acceleration

y_pred = regression.predict(x_test)
```

```python
#the points which are on the regression line is prdeicted perfectly
sns.regplot(x=y_test,y=y_pred)
plt.title("Actual MPG vs Predicted MPG")
```

```python
from sklearn.metrics import mean_absolute_error
```

```python
print("Mean Absolute Error = ",mean_absolute_error(y_test,y_pred))
```

```python
from sklearn.metrics import mean_squared_error
print("Mean Squared Error = ",mean_squared_error(y_test,y_pred))
```
