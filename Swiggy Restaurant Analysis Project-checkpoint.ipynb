```python
#Finding Avg Price on Swiggy of all Restaurant of All Over India
avg_price = swiggy['Price'].mean()
print("Average Price:",avg_price)
```

```python
#This shows the Numeber of Cities in Which Swiggy Service is there
city_count = swiggy.groupby('City').count()
city_count.index
```

```python
#Here This Shows the Number Of Order Id is Done by City Wise
res = swiggy.groupby('Area').count()
res['ID']
```

```python
mean_delivery_time = swiggy['Delivery time'].mean()
print("Mean Delivery time : ",mean_delivery_time)
```

```python
#These Scatter Plot shows that there is no correlation between Price and Avg Ratings
#As in high price restaurant the ratings are also not high
plt.scatter(x=swiggy['Price'],y=swiggy['Avg ratings'])
plt.title("Price vs Avg Ratings")
plt.xlabel("Price")
plt.ylabel("Avg Ratings")
plt.show()
```

```python
#here from this graph we can conclude that in which city the avg delivery time and 
#the outlier are show as the delivery time is long
sns.boxplot(x=swiggy['City'],y=swiggy['Delivery time'])
plt.title("City wise Delivery Analysis")
```

```python
#Here from these graph we can conclude that Kolkata has highest number of online food order
sns.countplot(x=swiggy['City'],data=swiggy)
plt.title("City vs Count")
plt.xlabel("City")
plt.ylabel("Count")
plt.show()
```

```python
#from the boxplot the median rating of avg rating is around 3.8
plt.boxplot(x=swiggy['Avg ratings'],data=swiggy)
```
